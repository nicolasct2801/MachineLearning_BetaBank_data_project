ahora nombramos 'target' al objetivo y 'features' a las caracteristicas.

```python
encoder = OneHotEncoder(drop='first', sparse=False)
encoded_features = encoder.fit_transform(df[['Geography', 'Gender']])
encoded_df = pd.DataFrame(encoded_features, columns=encoder.get_feature_names(['Geography', 'Gender']))

# Eliminar columnas originales y agregar las codificadas
df = df.drop(['Geography', 'Gender'], axis=1)
df = pd.concat([df, encoded_df], axis=1)
```

Muy bien Nicolas, has identificado que el dataset está desbalanceado. Esto es clave para definir el plan de trabajo

```python
features = df.drop(columns=['Exited'])
target = df['Exited']
```

ahora despues de dividir los datos en caracteristicas y objetivo, dividiremos los datos en conjunto de entrenamiento y prueba

Dividimos en entrenamiento, validación y prueba

```python
features_train, features_temp, target_train, target_temp = train_test_split(
    features, target, test_size=0.3, random_state=12345, stratify=target)
features_valid, features_test, target_valid, target_test = train_test_split(
    features_temp, target_temp, test_size=0.5, random_state=12345, stratify=target_temp)
```

Escalar los datos

```python
scaler = StandardScaler()
features_train = scaler.fit_transform(features_train)
features_valid = scaler.transform(features_valid)
features_test = scaler.transform(features_test)
```

Optimización de hiperparámetros para RandomForest

```python
param_grid_rf = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, 30]
}
gs_rf = GridSearchCV(RandomForestClassifier(random_state=12345, class_weight='balanced'),
                      param_grid_rf, scoring='f1', cv=3)
gs_rf.fit(features_train, target_train)
best_rf = gs_rf.best_estimator_
```

Entrenar el mejor modelo RandomForest

```python
best_rf.fit(features_train, target_train)
target_pred_rf = best_rf.predict(features_test)
f1_rf = f1_score(target_test, target_pred_rf)
roc_auc_rf = roc_auc_score(target_test, best_rf.predict_proba(features_test)[:, 1])

print(f'RandomForest - F1 Score: {f1_rf}')
print(f'RandomForest - AUC-ROC: {roc_auc_rf}')
```

Optimización de hiperparámetros para LogisticRegression

```python
param_grid_lr = {
    'C': [0.01, 0.1, 1, 10]
}
gs_lr = GridSearchCV(LogisticRegression(random_state=12345, class_weight='balanced'),
                      param_grid_lr, scoring='f1', cv=3)
gs_lr.fit(features_train, target_train)
best_lr = gs_lr.best_estimator_
```

Entrenar el mejor modelo LogisticRegression

```python
best_lr.fit(features_train, target_train)
target_pred_lr = best_lr.predict(features_test)
f1_lr = f1_score(target_test, target_pred_lr)
roc_auc_lr = roc_auc_score(target_test, best_lr.predict_proba(features_test)[:, 1])

print(f'LogisticRegression - F1 Score: {f1_lr}')
print(f'LogisticRegression - AUC-ROC: {roc_auc_lr}')
print('¡Entrenado!')
```
